11-— Find Geneformer's token dictionary

```python
geneformer_dir = "/content/Geneformer"

print("Searching for token dictionary files...\n")

matches = []

for root, dirs, files in os.walk(geneformer_dir):
    for file in files:
        name = file.lower()
        if any(x in name for x in [
            "token",
            "vocab",
            "gene_dict",
            "gene_dictionary"
        ]):
            matches.append(os.path.join(root, file))

for path in matches:
    print(path)

print("\nNumber of possible files found:", len(matches))
```

14-— Separate expression genes from metadata

```python
# Metadata columns
metadata_cols = ["Cancer_Type", "tissue_type"]

# Expression columns
gene_columns = [c for c in train.columns if c not in metadata_cols]

print("Total train columns:", len(train.columns))
print("Metadata columns:", len(metadata_cols))
print("Expression columns:", len(gene_columns))

print("\nMetadata:")
print(metadata_cols)

print("\nFirst 20 gene columns:")
print(gene_columns[:20])

print("\nLast 20 gene columns:")
print(gene_columns[-20:])
```

```python
# Separate numeric-looking IDs from gene symbols
numeric_gene_ids = []
symbol_gene_ids = []

for gene in gene_columns:
    gene_str = str(gene)

    if gene_str.isdigit():
        numeric_gene_ids.append(gene_str)
    else:
        symbol_gene_ids.append(gene_str)

print("Numeric/Entrez-like IDs:", len(numeric_gene_ids))
print("Symbol-like IDs:", len(symbol_gene_ids))

print("\nFirst 20 numeric IDs:")
print(numeric_gene_ids[:20])

print("\nFirst 20 symbols:")
print(symbol_gene_ids[:20])
```

# ***gene-ID mapping: steps:***

---



**Step 1 — Separate expression from labels**

```python
label_columns = ["Cancer_Type", "tissue_type"]

gene_columns = [
    c for c in train.columns
    if c not in label_columns
]

print("Number of expression columns:", len(gene_columns))
print("Labels:", label_columns)

print("\nFirst 20 expression columns:")
print(gene_columns[:20])

print("\nLast 20 expression columns:")
print(gene_columns[-20:])
```

**Step 2 — Load the Geneformer token dictionary**

```python
import pickle

token_dict_path = "/content/Geneformer/geneformer/token_dictionary_gc104M.pkl"

with open(token_dict_path, "rb") as f:
    token_dictionary = pickle.load(f)

print("Geneformer genes:", len(token_dictionary))

print(list(token_dictionary.items())[:10])
```

**Step 3 — Check whether data contains Ensembl IDs**

```python
ensembl_columns = [
    c for c in gene_columns
    if str(c).startswith("ENSG")
]

symbol_columns = [
    c for c in gene_columns
    if str(c)[0].isalpha()
]

numeric_columns = [
    c for c in gene_columns
    if str(c).isdigit()
]

print("Ensembl IDs:", len(ensembl_columns))
print("Symbol-like IDs:", len(symbol_columns))
print("Numeric/Entrez-like IDs:", len(numeric_columns))
```
